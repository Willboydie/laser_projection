```python
import shapes
import numpy as np

inc = 0.1
c = 200
r = 50

path = shapes.cartesianRectangle(inc, c, r)
rx_ratio = []
ry_ratio = []
rz_ratio = []

for i,point in enumerate(path):
    x, y, z = point
    r = np.sqrt(x**2 + y**2 + z**2)
    rx_ratio.append([x, x / r])
    ry_ratio.append([y, y / r])
    rz_ratio.append([z, z / r])

rx_ratio = np.array(rx_ratio)
ry_ratio = np.array(ry_ratio)
rz_ratio = np.array(rz_ratio)
```

```python
import matplotlib.pyplot as plt

# Plotting
fig, ax = plt.subplots(1, 3, figsize=(30, 10))
ax[0].plot(rx_ratio[:, 0], rx_ratio[:, 1])
ax[1].plot(ry_ratio[:, 0], ry_ratio[:, 1])
ax[2].plot(rz_ratio[:, 0], rz_ratio[:, 1])
ax[0].set_xlabel('x')
ax[0].set_ylabel('x / r')
ax[1].set_xlabel('y')
ax[1].set_ylabel('y / r')
ax[2].set_xlabel('z')
ax[2].set_ylabel('z / r')

plt.show()
```
